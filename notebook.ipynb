## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana perkembangan total nilai penjualan dan jumlah pesanan Olist setiap bulan selama periode September 2016 hingga Oktober 2018, serta pada bulan apa terjadi peningkatan atau penurunan penjualan paling signifikan sebagai dasar evaluasi strategi penjualan?
- **Pertanyaan 2:** Bagaimana perbedaan rata-rata skor ulasan pelanggan (review score) antara pesanan yang dikirim tepat waktu dan terlambat selama periode September 2016 hingga Oktober 2018, serta bagaimana hasilnya dapat digunakan untuk mengevaluasi dan meningkatkan kualitas layanan pengiriman Olist?
- **Pertanyaan 3:** Kategori produk apa saja yang memiliki kontribusi nilai penjualan dan jumlah pesanan tertinggi selama periode September 2016 hingga Oktober 2018, serta bagaimana hasil analisis tersebut dapat digunakan untuk menentukan prioritas pengelolaan produk dan strategi promosi?

### Cleaning Data

## Exploratory Data Analysis (EDA)

### Penggabungan Data dan Persiapan Analisis

```python
# PERTANYAAN BISNIS 1-Analisis Perkembangan Penjualan dan Jumlah Pesanan Bulanan
# Mendefinisikan konstanta tanggal analisis yang belum terdefinisi
START_DATE = '2016-09-01'
END_DATE = '2018-11-01'
print("=" * 65)
print("EDA PERTANYAAN 1: ANALISIS PENJUALAN BULANAN")
print("=" * 65)
# 1. Mengambil pesanan yang berhasil diselesaikan
delivered_orders = orders_df[
    orders_df['order_status'] == 'delivered'
].copy()
# 2. Memfilter periode September 2016 - Oktober 2018
delivered_orders = delivered_orders[
    (delivered_orders['order_purchase_timestamp'] >= START_DATE) &
    (delivered_orders['order_purchase_timestamp'] < END_DATE)
].copy()
# 3. Menggabungkan data pesanan dan item produk
sales_data = pd.merge(
    delivered_orders,
    order_items_df,
    on='order_id',
    how='inner'
)
# 4. Menampilkan gambaran awal data penjualan
print("\n5 BARIS PERTAMA DATA PENJUALAN:")
display(sales_data.head())
# 5. Mengelompokkan data berdasarkan bulan
monthly_sales = sales_data.groupby(
    sales_data['order_purchase_timestamp'].dt.to_period('M')
).agg(
    total_sales=('price', 'sum'),
    total_orders=('order_id', 'nunique')
)
# 6. Melengkapi periode bulan agar tidak ada bulan terlewat
all_months = pd.period_range(
    start='2016-09',
    end='2018-10',
    freq='M'
)
monthly_sales = monthly_sales.reindex(
    all_months,
    fill_value=0
)
# 7. Mengubah index periode menjadi kolom
monthly_sales.index.name = 'month'
monthly_sales = monthly_sales.reset_index()
# 8. Mengubah format periode menjadi tanggal
monthly_sales['month'] = monthly_sales['month'].dt.to_timestamp()
# 9. Menghitung perubahan nilai penjualan
monthly_sales['sales_change'] = (
    monthly_sales['total_sales'].diff()
)
# 10. Menghitung persentase pertumbuhan penjualan
previous_sales = monthly_sales['total_sales'].shift(1)
monthly_sales['sales_growth'] = np.where(
    previous_sales > 0,
    (monthly_sales['total_sales'] - previous_sales) /
    previous_sales * 100,
    np.nan
)
# 11. Menghitung perubahan jumlah pesanan
monthly_sales['orders_change'] = (
    monthly_sales['total_orders'].diff()
)
# 12. Menampilkan seluruh hasil analisis penjualan bulanan
print("\nTABEL PERKEMBANGAN PENJUALAN DAN JUMLAH PESANAN:")
display(monthly_sales.round(2))
# 13. Mengidentifikasi peningkatan penjualan terbesar
highest_increase = monthly_sales.loc[
    monthly_sales['sales_change'].idxmax()
]
# 14. Mengidentifikasi penurunan penjualan terbesar
highest_decrease = monthly_sales.loc[
    monthly_sales['sales_change'].idxmin()
]
# 15. Menampilkan hasil peningkatan penjualan terbesar
print("\nBULAN DENGAN PENINGKATAN PENJUALAN TERBESAR:")
display(highest_increase.to_frame(name='Value').round(2))
# 16. Menampilkan hasil penurunan penjualan terbesar
print("\nBULAN DENGAN PENURUNAN PENJUALAN TERBESAR:")
display(highest_decrease.to_frame(name='Value').round(2))


# PERTANYAAN BISNIS 2-Analisis Ketepatan Pengiriman dan Kepuasan Pelanggan
print("\n" + "=" * 65)
print("EDA PERTANYAAN 2: ANALISIS KETEPATAN PENGIRIMAN")
print("=" * 65)

# 1. Menggabungkan beberapa ulasan menjadi satu skor per pesanan
reviews_clean = order_reviews_df.groupby('order_id').agg(
    review_score=('review_score', 'mean')
).reset_index()

# 2. Memeriksa hasil agregasi ulasan
print("\nPEMERIKSAAN DATA ULASAN:")
print("Jumlah data ulasan sebelum agregasi:",
      order_reviews_df.shape[0])
print("Jumlah data ulasan setelah agregasi:",
      reviews_clean.shape[0])
print("Jumlah duplikat order_id:",
      reviews_clean['order_id'].duplicated().sum())

print("\n5 BARIS PERTAMA DATA ULASAN:")
display(reviews_clean.head())

# 3. Mengambil pesanan delivered dengan tanggal pengiriman lengkap
# dan sesuai periode analisis
delivery_data = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_delivered_customer_date'].notna()) &
    (orders_df['order_estimated_delivery_date'].notna()) &
    (orders_df['order_purchase_timestamp'] >= START_DATE) &
    (orders_df['order_purchase_timestamp'] < END_DATE)
].copy()

# 4. Menggabungkan data pesanan dengan data ulasan
delivery_data = pd.merge(
    delivery_data,
    reviews_clean,
    on='order_id',
    how='inner'
)

# 5. Menghitung selisih hari pengiriman aktual dan estimasi
delivery_data['delivery_difference'] = (
    delivery_data['order_delivered_customer_date'] -
    delivery_data['order_estimated_delivery_date']
).dt.total_seconds() / 86400

# 6. Mengelompokkan status pengiriman
delivery_data['delivery_status'] = delivery_data[
    'delivery_difference'
].apply(
    lambda x: 'Tepat Waktu' if x <= 0 else 'Terlambat'
)

# 7. Menghitung rata-rata review score berdasarkan status pengiriman
review_comparison = delivery_data.groupby(
    'delivery_status'
)['review_score'].mean().reindex(
    ['Tepat Waktu', 'Terlambat']
)

# 8. Menampilkan hasil perbandingan review
print("\nRATA-RATA REVIEW SCORE BERDASARKAN STATUS PENGIRIMAN:")
display(review_comparison.round(2))

# 9. Menghitung jumlah pesanan berdasarkan status pengiriman
delivery_counts = delivery_data[
    'delivery_status'
].value_counts().reset_index()

delivery_counts.columns = [
    'delivery_status',
    'total_orders'
]

# 10. Menampilkan jumlah pesanan berdasarkan status pengiriman
print("\nJUMLAH PESANAN BERDASARKAN STATUS PENGIRIMAN:")
display(delivery_counts)

# 11. Menampilkan ringkasan data pengiriman
print("\nRINGKASAN DATA PENGIRIMAN:")
print("Total pesanan dianalisis:",
      delivery_data['order_id'].nunique())

print("Total pesanan tepat waktu:",
      delivery_data[
          delivery_data['delivery_status'] == 'Tepat Waktu'
      ]['order_id'].nunique())

print("Total pesanan terlambat:",
      delivery_data[
          delivery_data['delivery_status'] == 'Terlambat'
      ]['order_id'].nunique())

# PERTANYAAN BISNIS 3=Analisis Kontribusi Penjualan dan Jumlah Pesanan
# Berdasarkan Kategori Produk

print("\n" + "=" * 65)
print("EDA PERTANYAAN 3: ANALISIS KATEGORI PRODUK")
print("=" * 65)

# 1. Mengambil ID pesanan delivered sesuai periode analisis
delivered_order_ids = orders_df.loc[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_purchase_timestamp'] >= START_DATE) &
    (orders_df['order_purchase_timestamp'] < END_DATE),
    ['order_id']
]

# 2. Menggabungkan data item dengan pesanan delivered
product_sales = pd.merge(
    order_items_df,
    delivered_order_ids,
    on='order_id',
    how='inner'
)

# 3. Menggabungkan data item dengan informasi produk
product_sales = pd.merge(
    product_sales,
    products_df[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# 4. Menggabungkan dengan tabel terjemahan kategori
product_sales = pd.merge(
    product_sales,
    category_translation_df,
    on='product_category_name',
    how='left'
)

# 5. Mengatasi kategori produk yang tidak memiliki terjemahan
product_sales['product_category_name_english'] = (
    product_sales['product_category_name_english']
    .fillna('Unknown')
)

# 6. Mengelompokkan berdasarkan kategori produk
category_sales = product_sales.groupby(
    'product_category_name_english'
).agg(
    total_sales=('price', 'sum'),
    total_orders=('order_id', 'nunique'),
    total_items=('order_item_id', 'count')
).reset_index()

# 7. Mengurutkan kategori berdasarkan nilai penjualan
category_sales = category_sales.sort_values(
    by='total_sales',
    ascending=False
).reset_index(drop=True)

# 8. Menampilkan seluruh hasil analisis kategori produk
print("\nTABEL KONTRIBUSI PENJUALAN DAN PESANAN SETIAP KATEGORI:")
display(category_sales.round(2))

# 9. Mengambil 10 kategori dengan nilai penjualan tertinggi
top_sales = category_sales.sort_values(
    by='total_sales',
    ascending=False
).head(10)

print("\n10 KATEGORI DENGAN NILAI PENJUALAN TERTINGGI:")
display(top_sales.round(2))

# 10. Mengambil 10 kategori dengan jumlah pesanan tertinggi
top_orders = category_sales.sort_values(
    by='total_orders',
    ascending=False
).head(10)

print("\n10 KATEGORI DENGAN JUMLAH PESANAN TERTINGGI:")
display(top_orders.round(2))

```

## Analisis Lanjutan

```python
# Mengambil pesanan yang berhasil dikirim
rfm_orders = orders_df[
    orders_df['order_status'] == 'delivered'
].copy()

# Menggabungkan data pesanan dengan item transaksi
rfm_data = pd.merge(
    rfm_orders,
    order_items_df,
    on='order_id',
    how='inner'
)

# Menggabungkan dengan data pelanggan untuk mendapatkan customer_unique_id
rfm_data = pd.merge(
    rfm_data,
    customers_df[['customer_id', 'customer_unique_id']],
    on='customer_id',
    how='left'
)

# Menentukan tanggal acuan analisis
snapshot_date = (
    rfm_data['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

# Menghitung RFM setiap pelanggan
rfm = rfm_data.groupby('customer_unique_id').agg(
    Recency=(
        'order_purchase_timestamp',
        lambda x: (snapshot_date - x.max()).days
    ),
    Frequency=('order_id', 'nunique'),
    Monetary=('price', 'sum')
).reset_index()

# Menampilkan hasil RFM
display(rfm.head())

# Recency: semakin kecil semakin baik
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Frequency: semakin besar semakin baik
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Monetary: semakin besar semakin baik
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Menghitung total skor RFM
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

display(rfm.head())
def segment_customer(score):
    if score >= 10:
        return 'Champions'
    elif score >= 8:
        return 'Loyal Customers'
    elif score >= 6:
        return 'Potential Customers'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Others'

rfm['Customer_Segment'] = rfm['RFM_Score'].apply(
    segment_customer
)

# Melihat hasil segmentasi pelanggan
display(
    rfm.groupby('Customer_Segment').agg(
        Total_Customers=('customer_unique_id', 'count'),
        Average_Recency=('Recency', 'mean'),
        Average_Frequency=('Frequency', 'mean'),
        Average_Monetary=('Monetary', 'mean')
    ).round(2)
)
# Menghitung jumlah pelanggan setiap segmen
segment_counts = rfm['Customer_Segment'].value_counts()

# Visualisasi 1: Distribusi pelanggan berdasarkan segmen RFM
plt.figure(figsize=(10, 6))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values
)

plt.title('Distribusi Pelanggan Berdasarkan Segmen RFM')
plt.xlabel('Customer Segment')
plt.ylabel('Jumlah Pelanggan')
plt.xticks(rotation=20)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# Visualisasi 2: Distribusi nilai pembelian berdasarkan segmen RFM
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=rfm,
    x='Customer_Segment',
    y='Monetary'
)

plt.title('Distribusi Nilai Pembelian Berdasarkan Segmen RFM')
plt.xlabel('Customer Segment')
plt.ylabel('Total Nilai Pembelian Produk (BRL)')
plt.xticks(rotation=20)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Insight:**

Berdasarkan hasil analisis RFM, segmen Potential Customers dan Loyal Customers memiliki jumlah pelanggan terbanyak. Hal ini menunjukkan bahwa sebagian besar pelanggan memiliki potensi untuk menjadi pelanggan loyal. Segmen Champions memiliki jumlah lebih sedikit, sedangkan Others merupakan segmen dengan jumlah pelanggan paling rendah.

Pada visualisasi nilai pembelian, segmen Champions menunjukkan nilai pembelian yang relatif lebih tinggi dibandingkan segmen lainnya. Distribusi pembelian pada setiap segmen juga memiliki banyak pencilan (outlier), yang menunjukkan adanya pelanggan dengan nilai transaksi jauh lebih besar dibandingkan pelanggan lainnya.

Hasil analisis ini menunjukkan bahwa Olist perlu mempertahankan pelanggan Champions melalui program loyalitas, meningkatkan keterlibatan Potential Customers agar menjadi pelanggan loyal, serta melakukan promosi khusus untuk mengaktifkan kembali pelanggan At Risk.

Dengan demikian, RFM Analysis dapat membantu Olist memahami karakteristik pelanggan dan menentukan strategi pemasaran yang lebih tepat berdasarkan perilaku pembelian masing-masing segmen.
