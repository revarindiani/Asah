## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana perkembangan total nilai penjualan dan jumlah pesanan Olist setiap bulan selama periode September 2016 hingga Oktober 2018, serta pada bulan apa terjadi peningkatan atau penurunan penjualan paling signifikan sebagai dasar evaluasi strategi penjualan?
- **Pertanyaan 2:** Bagaimana perbedaan rata-rata skor ulasan pelanggan (review score) antara pesanan yang dikirim tepat waktu dan terlambat selama periode September 2016 hingga Oktober 2018, serta bagaimana hasilnya dapat digunakan untuk mengevaluasi dan meningkatkan kualitas layanan pengiriman Olist?
- **Pertanyaan 3:** Kategori produk apa saja yang memiliki kontribusi nilai penjualan dan jumlah pesanan tertinggi selama periode September 2016 hingga Oktober 2018, serta bagaimana hasil analisis tersebut dapat digunakan untuk menentukan prioritas pengelolaan produk dan strategi promosi?

### Cleaning Data

#### Fixing ... problem

```python
# Menghapus data duplikat
geolocation_df.drop_duplicates(inplace=True)
# Memeriksa hasil cleaning
print("Jumlah duplikat:", geolocation_df.duplicated().sum())

# Mengisi missing values pada kategori produk
products_df['product_category_name'] = products_df['product_category_name'].fillna('Unknown')
# Memeriksa hasil cleaning
print("Missing values setelah cleaning:")
print(products_df.isna().sum())

# Mengubah tipe data kolom tanggal
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for column in date_columns:
    orders_df[column] = pd.to_datetime(orders_df[column], errors='coerce')
# Memeriksa hasil cleaning
orders_df.info()

# Mengubah tipe data shipping_limit_date
order_items_df['shipping_limit_date'] = pd.to_datetime(
    order_items_df['shipping_limit_date'],
    errors='coerce'
)
# Memeriksa hasil cleaning
order_items_df.info()

# Mengubah tipe data kolom tanggal ulasan
order_reviews_df['review_creation_date'] = pd.to_datetime(
    order_reviews_df['review_creation_date'],
    errors='coerce'
)
order_reviews_df['review_answer_timestamp'] = pd.to_datetime(
    order_reviews_df['review_answer_timestamp'],
    errors='coerce'
)
# Memeriksa hasil cleaning
order_reviews_df.info()

# Menggabungkan beberapa ulasan menjadi satu skor per pesanan
reviews_clean = order_reviews_df.groupby('order_id').agg(
    review_score=('review_score', 'mean')
).reset_index()

# Memeriksa hasil
print("Jumlah data ulasan sebelum cleaning:", order_reviews_df.shape[0])
print("Jumlah data ulasan setelah cleaning:", reviews_clean.shape[0])
print("Jumlah duplikat order_id:", reviews_clean['order_id'].duplicated().sum())

reviews_clean.head()

# Memeriksa kondisi seluruh dataset setelah cleaning
datasets = {
    'Customers': customers_df,
    'Geolocation': geolocation_df,
    'Order Items': order_items_df,
    'Order Payments': order_payments_df,
    'Order Reviews': order_reviews_df,
    'Orders': orders_df,
    'Products': products_df,
    'Sellers': sellers_df,
    'Category Translation': category_translation_df
}
for name, df in datasets.items():
    print(f"\n===== {name} =====")
    print("Shape:", df.shape)
    print("Total Missing Values:", df.isna().sum().sum())
    print("Total Duplicates:", df.duplicated().sum())
```

**Insight:**

Pada dataset Products, sebanyak 610 missing values pada kolom `product_category_name` diisi dengan nilai `Unknown`, sedangkan missing values pada atribut numerik tetap dipertahankan. Selanjutnya, kolom tanggal pada dataset Orders, Order Items, dan Order Reviews dikonversi dari tipe object menjadi datetime agar dapat digunakan dalam analisis berbasis waktu. Missing values pada kolom tanggal pesanan dan komentar ulasan tetap dipertahankan karena berkaitan dengan kondisi pesanan serta sifat komentar yang opsional. ## Exploratory Data Analysis (EDA)

### Explore ...

```python
# Mengambil pesanan yang berhasil diselesaikan
delivered_orders = orders_df[
    orders_df['order_status'] == 'delivered'
].copy()

# Menggabungkan data orders dan order items
sales_data = pd.merge(
    delivered_orders,
    order_items_df,
    on='order_id',
    how='inner'
)
# Memfilter periode September 2016 - Oktober 2018
sales_data = sales_data[
    (sales_data['order_purchase_timestamp'] >= '2016-09-01') &
    (sales_data['order_purchase_timestamp'] < '2018-11-01')
]

# Menampilkan lima baris pertama
sales_data.head()

# Mengelompokkan data berdasarkan bulan
monthly_sales = sales_data.groupby(
    sales_data['order_purchase_timestamp'].dt.to_period('M')
).agg(
    total_sales=('price', 'sum'),
    total_orders=('order_id', 'nunique')
)

# Melengkapi periode bulan agar tidak ada bulan yang terlewat
all_months = pd.period_range(
    start='2016-09',
    end='2018-10',
    freq='M'
)

monthly_sales = monthly_sales.reindex(all_months, fill_value=0)

# Mengubah index periode menjadi kolom
monthly_sales.index.name = 'month'
monthly_sales = monthly_sales.reset_index()

# Mengubah format periode menjadi tanggal
monthly_sales['month'] = monthly_sales['month'].dt.to_timestamp()

# Menampilkan hasil
monthly_sales

# Menghitung perubahan nilai penjualan
monthly_sales['sales_change'] = monthly_sales['total_sales'].diff()

# Menghitung persentase pertumbuhan penjualan
monthly_sales['sales_growth'] = (
    monthly_sales['total_sales'].pct_change() * 100
)

# Menghitung perubahan jumlah pesanan
monthly_sales['orders_change'] = monthly_sales['total_orders'].diff()

# Menampilkan hasil analisis
monthly_sales

# Menggabungkan data penjualan dengan data produk untuk mendapatkan kategori
product_sales = pd.merge(
    sales_data,
    products_df[['product_id', 'product_category_name']],
    on='product_id',
    how='inner'
)

# Menggabungkan dengan tabel terjemahan kategori ke bahasa Inggris
product_sales = pd.merge(
    product_sales,
    category_translation_df,
    on='product_category_name',
    how='inner'
)

# Mengelompokkan berdasarkan kategori produk bahasa Inggris dan menghitung total penjualan
top_sales = product_sales.groupby(
    'product_category_name_english'
).agg(
    total_sales=('price', 'sum')
).reset_index()

top_sales = top_sales.sort_values(
    by='total_sales',
    ascending=False
).head(10)
```

```python
# Mencari bulan dengan kenaikan penjualan terbesar
highest_increase = monthly_sales.loc[
    monthly_sales['sales_change'].idxmax()
]

# Mencari bulan dengan penurunan penjualan terbesar
highest_decrease = monthly_sales.loc[
    monthly_sales['sales_change'].idxmin()
]

print("Kenaikan Penjualan Terbesar:")
print(highest_increase)

print("\nPenurunan Penjualan Terbesar:")
print(highest_decrease)
```

### Pertanyaan 2:

```python
# Mengambil data pesanan delivered
delivery_data = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_delivered_customer_date'].notna()) &
    (orders_df['order_estimated_delivery_date'].notna())
].copy()

# Menggabungkan dengan review
delivery_data = delivery_data.merge(
    reviews_clean,
    on='order_id',
    how='inner'
)

# Menghitung selisih waktu pengiriman
delivery_data['delivery_difference'] = (
    delivery_data['order_delivered_customer_date'] -
    delivery_data['order_estimated_delivery_date']
).dt.total_seconds() / 86400

# Mengelompokkan status pengiriman
delivery_data['delivery_status'] = delivery_data[
    'delivery_difference'
].apply(
    lambda x: 'Tepat Waktu' if x <= 0 else 'Terlambat'
)

delivery_data.head()

review_comparison = delivery_data.groupby(
    'delivery_status'
).agg(
    average_review=('review_score', 'mean')
).reset_index()

plt.figure(figsize=(9, 6))

sns.barplot(
    data=review_comparison,
    x='delivery_status',
    y='average_review'
)

plt.title(
    'Rata-rata Review Berdasarkan Ketepatan Pengiriman',
    fontsize=15,
    fontweight='bold'
)
plt.xlabel('Status Pengiriman')
plt.ylabel('Rata-rata Review Score')
plt.ylim(0, 5)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()

delivery_counts = delivery_data[
    'delivery_status'
].value_counts().reset_index()

delivery_counts.columns = [
    'delivery_status',
    'total_orders'
]

plt.figure(figsize=(9, 6))

sns.barplot(
    data=delivery_counts,
    x='delivery_status',
    y='total_orders'
)

plt.title(
    'Jumlah Pesanan Berdasarkan Status Pengiriman',
    fontsize=15,
    fontweight='bold'
)
plt.xlabel('Status Pengiriman')
plt.ylabel('Jumlah Pesanan')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

```python
# Mengambil pesanan yang berhasil dikirim
rfm_orders = orders_df[
    orders_df['order_status'] == 'delivered'
].copy()

# Menggabungkan data pesanan dengan item transaksi
rfm_data = pd.merge(
    rfm_orders,
    order_items_df,
    on='order_id',
    how='inner'
)

# Menggabungkan dengan data pelanggan
rfm_data = pd.merge(
    rfm_data,
    customers_df[['customer_id', 'customer_unique_id']],
    on='customer_id',
    how='left'
)

rfm_data.head()
# Menentukan tanggal acuan analisis
snapshot_date = (
    rfm_data['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

# Menghitung RFM
rfm = rfm_data.groupby('customer_unique_id').agg(
    Recency=(
        'order_purchase_timestamp',
        lambda x: (snapshot_date - x.max()).days
    ),
    Frequency=('order_id', 'nunique'),
    Monetary=('price', 'sum')
).reset_index()

rfm.head()
# Membagi pelanggan menjadi 4 kelompok skor
# Recency: semakin kecil semakin baik
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Frequency: semakin besar semakin baik
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Monetary: semakin besar semakin baik
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Menghitung total skor
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

rfm.head()
def segment_customer(score):
    if score >= 10:
        return 'Champions'
    elif score >= 8:
        return 'Loyal Customers'
    elif score >= 6:
        return 'Potential Customers'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Others'

rfm['Customer_Segment'] = rfm['RFM_Score'].apply(
    segment_customer
)

rfm.head()

import matplotlib.pyplot as plt
import seaborn as sns

segment_counts = rfm['Customer_Segment'].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=segment_counts.index,
    y=segment_counts.values
)

plt.title('Distribusi Pelanggan Berdasarkan Segmen RFM')
plt.xlabel('Customer Segment')
plt.ylabel('Jumlah Pelanggan')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=rfm,
    x='Customer_Segment',
    y='Monetary'
)

plt.title('Distribusi Nilai Pembelian Berdasarkan Segmen RFM')
plt.xlabel('Customer Segment')
plt.ylabel('Total Pembelian')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

- Others merupakan segmen dengan jumlah pelanggan paling sedikit.
Artinya, sebagian besar pelanggan berada pada segmen Potential Customers dan Loyal Customers. Hal ini menunjukkan adanya peluang untuk meningkatkan frekuensi pembelian pelanggan agar lebih banyak yang berkembang menjadi Champions.
